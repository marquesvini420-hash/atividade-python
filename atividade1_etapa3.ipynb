#Questão 5

```python
nota1= int(input("digite sua nota:"))
nota2 = int(input("digite a sua segunda nota:"))
notamedia = (nota1 + nota2) / 2
if notamedia >= 7:
  print("aprovado")
elif 7 > notamedia >= 5:
  print("recuperação")
else:
  print("reprovado")
```
