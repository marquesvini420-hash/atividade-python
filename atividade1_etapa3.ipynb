#Questão 14

```python
numerosecreto = 7
tentativas = 0

while True:
  palpite = int(input("tente descobrir o numero secreto:"))
  tentativas += 1
  if palpite == numerosecreto:
     print("voce acertou o numero secreto")
     print(f"numero de tentativas:{tentativas}")
     break
  elif palpite < numerosecreto:
    print("tente um numero maior")
  else:
    print("tente um numero menor")
```
