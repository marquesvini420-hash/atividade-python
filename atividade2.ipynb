#5. Máquina de venda de ingresso


```python
valoringresso = 25
total = 0
while True:
  valor = int(input("ensira o dinheiro:"))
  total += valor

  if  total >= valoringresso:
    break

print("total:",total)
print("troco:",total - valoringresso)
```

#10.Caixa eletrônico simplificado

```python
valr = int(input("Digite o valor: "))

cedulas = [100, 50, 20, 10, 5, 2, 1]
totalcedulas = 0

for c in cedulas:
    quantidade = valr // c
    valr = valr % c
    totalcedulas += quantidade

print("Quantidade mínima de cédulas:", totalcedulas)
```

#12.Divisores de um número

```python
n = int(input("digite um numero: "))

quantidad = 0

for d in range(1, n + 1):
  if n % d == 0:
    print(d)
    quantidad += 1
print("quantidade de divisores:", quantidad)
```

#14.Jogo com limite de tentativas

```python
secreto = 37
tentativas = 0

while tentativas < 5:
    palpite = int(input("Digite seu palpite: "))
    tentativas += 1

    if palpite == secreto:
        print("Você acertou!")
        break
    elif palpite < secreto:
        print("O número secreto é maior")
    else:
        print("O número secreto é menor")

else:
    print("Você usou todas as tentativas. O número era", secreto)
```
