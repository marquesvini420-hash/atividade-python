```python
listatemp = []
for d in range(6):
    num = int(input("Digite um número inteiro: "))
    listatemp.append(num)

numeros = tuple(listatemp)
cinco = 0
pares = 0
pcinco = -1
for i in range(len(numeros)):
    if numeros[i] == 5:
        cinco += 1
        if pcinco == -1:
            pcinco = i
    if numeros[i] % 2 == 0:
        pares += 1
print("Tupla:", numeros)
print("Quantidade de vezes que o 5 apareceu:", cinco)
if pcinco != -1:
    print("Posição da primeira ocorrência do 5:", pcinco)
else:
    print("O número 5 não foi digitado.")

print("Quantidade de números pares:", pares)
```

#11


```python
contagem = {}

print("Digite 8 palavras:")
for e in range(8):
    palavra = input("Palavra " + str(e + 1) + ": ").lower()
    if palavra in contagem:
        contagem[palavra] += 1
    else:
        contagem[palavra] = 1
print("Frequência das palavras:")
for palavra, qtd in contagem.items():
    print(palavra + ": " + str(qtd) + " vezes")
```

#12

```python
estoque = {"pernil": 32,"picanha": 10,"patinho": 70,"bananhinha": 68
}
produto = input("Nome do produto vendido: ").lower()
vendidas = int(input("Quantidade vendida: "))

if produto in estoque:
    if estoque[produto] > vendidas:
        estoque[produto] -= vendidas
        print("Venda realizada com sucesso")
    else:
        print("Estoque insuficiente")
else:
    print("Produto não cadastrado")

print("Estoque final: ", estoque)
```

```python
catalogo = []
print("Cadastro de 5 Produto")
for i in range(5):
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    prod_tupla = (codigo, nome, preco)
    catalogo.append(prod_tupla)
busca_codigo = int(input("\nDigite o código do produto para buscar: "))
achou = False
for prod in catalogo:
    if prod[0] == busca_codigo:
        print("Produto encontrado:")
        print("Nome: ", prod[1])
        print("Preço: ", prod[2])
        achou = True
        break
if not achou:
    print("Código não encontrado no catálogo.")
```
